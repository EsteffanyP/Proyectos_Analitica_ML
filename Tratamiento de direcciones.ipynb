## <font color='8EC044'> **Paso 4:** Gestionar variables

### **Variable `fecha`**



Se opta por eliminar la hora que está luego la fecha porque no aporta información relevante.

```python
df['fecha'] = df['fecha'].apply(lambda x: pd.to_datetime(x).date() if x != 'NR' else x) # Ignora las entradas 'NR'.
df['fecha']
```

### **Variable `razon social`**

Para estandarizar el formato se convierten todos los nombres a mayúsculas y se eliminan los posibles saltos de línea.

También se hacen reemplazos para escribir correctamente las abreviaturas de los tipos de sociedades.

```python
df['razon social'] = df['razon social'].str.replace('\n', ' ', regex=True)  # Elimina saltos de línea.
df['razon social'] = df['razon social'].str.upper()  # Convierte todo a mayúsculas.
df['razon social'] = df['razon social'].str.replace(r'\.', '', regex=True) # Elimina todos los puntos para luego agregarlos correctamente.
df['razon social'] = df['razon social'].replace(r'\bSAS\b', 'S.A.S.', regex=True)
df['razon social'] = df['razon social'].replace(r'\bSA\b', 'S.A.', regex=True)
df['razon social'] = df['razon social'].replace(r'\bESP\b', 'E.S.P.', regex=True)
df['razon social'] = df['razon social'].replace(r'\bBIC\b', 'BIC.', regex=True)
df['razon social']
```

### **Variable `NIT`**

Para gestionar esta variable se tiene en cuenta que en Colombia los NIT pueden tener 9 o 10 dígitos. Conociendo ese estándar, muchos sitios web están adaptados para recibir solo los números. Es así, que se eliminan todos los caracteres distintos a números.

```python
def estandarizar_nit(nit): # Función que elimina los puntos, las comas y guiones.
    nit = str(nit)
    return nit.replace('.', '').replace('-', '').replace(' ', '')
df['NIT'] = df['NIT'].apply(estandarizar_nit)
df['NIT']
```

### **Variable `representante`**

Se quiere que los nombres queden en un solo formato, para esto se deciden que los nombres estén en mayúsculas.

```python
def estandarizar_nombre(nombre): # Se crea una función que estandarice variables con nombres.
    nombre = str(nombre)
    nombre = nombre.replace("\n", " ")  # Elimina los saltos de línea.
    return nombre.upper()  # Convierte a mayúsculas.
df['representante'] = df['representante'].apply(estandarizar_nombre)
df['representante']
```

### **Variable `direccion`**

Para estandarizar las direcciones se tiene en cuenta la [Propuesta de Estándar de las Direcciones
Urbanas para los Equipamientos del Ministerio de Educación](https://www.mineducacion.gov.co/1621/articles-193290_estandar_direcciones_urbanas.pdf). Este documento establece abreviaturas para los elementos que componen la dirección, además de lo siguiente:


* No se tendrán en cuenta la palabra “número” o cualquiera de sus abreviaturas (No.,
NUM., N., NO., etc.).
* No se utilizará ningún signo de puntuación.
* Cuando se haga actualización y/o cambios de nomenclatura, se puede tener 2 o más direcciones diferentes para un mismo predio, en estos casos se debe especificar si la dirección capturada corresponde con la nueva nomenclatura o la antigua.

El formato que propone también considera espacios entre numeros de calle y letras acompañantes, por ejemplo: "51A" pasa a ser "51 A".


```python
!pip install unidecode
from unidecode import unidecode
import re # Para expresiones regulares
df.loc[:, 'direccion'] = df['direccion'].apply(unidecode) # Elimina los acentos
df.loc[:, 'direccion'] = df['direccion'].str.replace(r'[.,°—\-:#]', ' ', regex=True) # Elimina los caracteres .,°—\-:#
df.loc[:, 'direccion'] = df['direccion'].str.replace(r'(\d)([A-Za-z])', r'\1 \2', regex=True) # Separar números de las letras
df.loc[:, 'direccion'] = df['direccion'].str.replace(r'\s+', ' ', regex=True)  # Elimina los espacios en blanco
df.loc[:, 'direccion'] = df['direccion'].str.upper() # Convierte a mayúsculas.
df.loc[:, 'direccion'] = df['direccion'].str.replace(r'\bNO\b|\bNDEG\b|\bNODEG\b|\bNRO\b', '', regex=True) # Elimina los registros de NO (correspondientes a calles o carreras).
```

Ahora, se buscan las palabras únicas en la columna `direccion` para encontrar los elementos de dirección utilizados en la base de datos.

```python
texto = ' '.join(df['direccion']) # Junta todas las palabras de la columna.
palabras = texto.split() # Separa en palabras.
palabras_no_numeros = [palabra for palabra in palabras if not any(char.isdigit() for char in palabra)] # Excluye las palabras que contengan números.
palabras_ordenadas = sorted(set(palabras_no_numeros)) # Ordena las palabras alfabéticamente.
for palabra in palabras_ordenadas:
    print(palabra)
```

Manualmente se identifican los reemplazos que se deben realizar y se escriben en un diccionario.

```python
reemplazos = {
    "AVENIDA CALLE": "AC", # Se coloca primero para evitar que se reemplace incorrectamente como "AV CL".
    "AVENIDA CARRERA": "AK",  # Se coloca primero para evitar que se reemplace incorrectamente como "AV KR".
    "AEROPUERTO": "AER",
    "AUTOPISTA": "AU",
    "AVENIDA": "AV",
    "BARRIO": "BR",
    "BOD": "BG", "BODEGA": "BG", "BODEGAS": "BG",
    "CALLE": "CL",
    "CARRERA": "KR", "CARRRERA": "KR", "CRA": "KR", "K": "KR",
    "CARRETERA": "CR",
    "CIUDADLE": "CIUDELA",
    "CORREGIMIENTO": "CORREG",
    "DIAG": "DG", "DIAGONAL": "DG",
    "ENTRADA": "EN",
    "ETAPA": "ET",
    "FINCA": "FI", "FCA": "FI",
    "INTERIOR": "IN",
    "KILOMETRO": "KM",
    "LOC": "L",
    "LOTE": "LT", "L": "LT",
    "MANZANA": "MZ",
    "OCCIDENTE": "OESTE",
    "PARQ": "PQ", "PARQUE": "PQ",
    "PLANTA": "PL",
    "PORTERIA": "PR",
    "PREDIO": "PD",
    "SECTOR": "SC",
    "TROCAL": "TC", "TRONCAL": "TC",
    "VEREDA": "VDA", "VRDA": "VDA",
    "VIA": "VI",
    "ZNA": "ZN", "ZONA": "ZN"
}
def reemplazar_palabras(direccion): # Función para reemplazar las palabras de la columna 'direccion'.
    for clave, valor in reemplazos.items():
        direccion = re.sub(r'\b' + re.escape(clave) + r'\b', valor, direccion) # Usa una expresión regular para asegurar que solo reemplaza la palabra exacta.
    return direccion
df.loc[:, 'direccion'] = df['direccion'].apply(reemplazar_palabras)
df['direccion']
```

Es importante mencionar que se encuentran casos especiales que no se modificaron correctamente con las anteriores funciones, por tanto, se cambian manualmente.

```python
df.loc[:, 'direccion'] = df['direccion'].str.replace(r'\b(KM)(\d+)', r'\1 \2', regex=True) # Añade espacio entre "KM" y el número.
df.loc[:, 'direccion'] = df['direccion'].str.replace(r'\b(KM \d+)\s(\d+)\b', r'\1.\2', regex=True) # Reemplaza espacios entre dos números por número decimal (ej.: "KM 3 5" cambia a "KM 3.5")
df.loc[:, 'direccion'] = df['direccion'].str.replace(r'\bDOSPUNTOCINCO\b', '2.5', regex=True) # Cambia el texto por el número 2.5
df.loc[:, 'direccion'] = df['direccion'].str.replace(r'\bGALAPAATLANTICO\b', 'GALAPA ATLANTICO', regex=True) # Separa el nombre de los dos lugares.
df.loc[:, 'direccion'] = df['direccion'].str.replace(r'\bBSUR\b', 'B SUR', regex=True) # Separa la letra B de SUR.
df.loc[:, 'direccion'] = df['direccion'].str.replace(r'\bOCCIDENTE\b', 'OESTE', regex=True) # Cambia OCCIDENTE por el sinónimo OESTE.
df.loc[:, 'direccion'] = df['direccion'].str.replace(r'\bNO42\b', '42', regex=True)
df.loc[:, 'direccion'] = df['direccion'].str.replace(r'\bNO52\b', '52', regex=True)
df.loc[:, 'direccion'] = df['direccion'].str.replace(r'\bKM 19.20\b', 'KM 19', regex=True) # Se mencionaban dos kilómetros, entonces se conserva solo uno.
```

Adicionalmente, se sabe que también había existencia de datos faltantes con registros "NO" y "No", pero no se eliminaron antes para que no interfiera con la gestión de la variable `direccion`, ya que varias observaciones de esa columna cotenían esas expresiones.

```python
df = df.replace("NO", "NA")
df = df.replace("No", "NA")
```

### **Variable ` material`**

Se observan los materiales que hay en la base de datos.

```python
df['material'].unique()
```

Dada la cantidad de materiales, se decide clasificarlos en 6 columnas de acuerdo con las clasificaciones de la Resolución 1407 de 2018: `papel`, `carton`, `plastico`, `vidrio`, `metal` y `otros`.
La variable `otros` contiene los materiales especificados en la base de datos que no se pueden incluir en las anteriores clasificaciones, estos son: "otros", "multimateriales", "tetra pack" y "llantas".

```python
def clasificar_material(i, material):
    pattern = fr'\b{material}\w*\b' # Para encontrar el nombre del material, sin importar que esté contenido  en un plural
    return 1 if re.search(pattern, i, re.IGNORECASE) else 0 # Busca el nombre del material en cada fila, ignorando mayúsculas o tildes. Devuelve 1 si lo encuentra, 0 de lo contrario.

# Se crean las columnas de los 6 grupos (papel, carton, vidrio, plástico, metal).
df['papel'] = df['material'].apply(lambda i: clasificar_material(i, 'papel'))
df['carton'] = df['material'].apply(lambda i: clasificar_material(i, 'carton'))
df['plastico'] = df['material'].apply(lambda i: clasificar_material(i, 'plastico'))
df['vidrio'] = df['material'].apply(lambda i: clasificar_material(i, 'vidrio'))
df['metal'] = df['material'].apply(lambda i: clasificar_material(i, 'metal'))

# Se crea la columna otros.
otros_materiales = ['otros', 'multimateriales', 'tetra pack', 'llantas']
df['otros'] = df['material'].apply(lambda i: 1 if any(clasificar_material(i, x) for x in otros_materiales) else 0)

df[['material', 'papel', 'carton',  'plastico', 'vidrio', 'metal', 'otros']]
```

```python
df = df.drop('material', axis=1) # Se elimina la columna original.
```

### **Variable `telefono`**

Se tienen varios casos especiales que se deben corregir antes de aplicar funciones que generalicen, así se evitan errores de formato.

```python
df['telefono'] = df['telefono'].replace('035-6930062', '(035)69300623', regex=False)
df['telefono'] = df['telefono'].replace('573788900', ' 3788900', regex=False) #Asumiendo que 57 se refiere al prefijo internacional para Colombia.
df['telefono'] = df['telefono'].replace('7459574/79/90', '7459574(79/90)', regex=False)
df['telefono'] = df['telefono'].replace('4057000 EXT 43222', '4057000(43222)', regex=False)
df['telefono'] = df['telefono'].replace('604 3783636', '(604)3783636', regex=False)
```

Se aplican expresiones regulares para estandarizar el formato de números de teléfono y poder separarlos más adelante.

```python
df['telefono'] = df['telefono'].replace(r'\n', '-', regex=True) # Cambiar saltos de linea por '-'
df['telefono'] = df['telefono'].replace(r'/', '-', regex=True) # Cambiar '/' por '-'
df['telefono'] = df['telefono'].replace('3208654701-3123-447787', '3208654701-3123447787', regex=False) # Es un caso especial que surge por los reemplazos anteriores.
df['telefono'] = df['telefono'].replace(r'-+', '-', regex=True) # Elimina '--' generados por los cambios de las lineas 1 y 2.
df['telefono'] = df['telefono'].replace(r'\s+', '', regex=True) # Elimina los espacios
```

Se separan los números de teléfono en varias columnas (tantas columnas como cantidad en la celda con más números de teléfono).

```python
def estandarizar_telefonos(telefono):
    telefono = str(telefono)  # Convierte a cadena de texto para que no genere error.
    return telefono.split('-') # Separa los números considerando el guión como separador.

telefonos_separados = df['telefono'].apply(estandarizar_telefonos)
max_telefonos = telefonos_separados.apply(len).max()  # Número máximo de teléfonos en una fila.
for i in range(max_telefonos):
    df[f'telefono_{i+1}'] = telefonos_separados.apply(lambda x: x[i] if i < len(x) else 'NR')
df[['telefono', 'telefono_1', 'telefono_2', 'telefono_3']]
```

## <font color='8EC044'> **Paso 5:** Gestionar outliers

La base de datos solo tiene una variable cuantitativa y es `capacidad`.

Se crea un boxplot para observar los *outliers* de la columna.

```python
import matplotlib.pyplot as plt
import seaborn as sns # Para hacer el boxplot.

df2 = df.copy() # Se hace una copia para eliminar los datos 'NR' o 'NA' sin afectar las otras variables del df original.
df2['capacidad'] = pd.to_numeric(df2['capacidad'], errors='coerce') # Convierte 'NR' a NaN para que no haya errores.

plt.figure(figsize=(8,6))
sns.boxplot(x=df2['capacidad'], color='#8EC044')
plt.title("Boxplot de la variable 'capacidad'")
plt.xlabel('Capacidad')
plt.show()
```

En este contexto, los valores atípicos proveen información de interés.
Nos interesa conocer las empresas que tienen mayor capacidad de procesamiento, en otras palabras, el tratamiento de estos valores extremos no es la eliminación.

Para saber cuales son las empresas que tienen una capacidad de procesamiento "atípica" se utiliza el método del Rango Intercuartílico (RIC). Este arroja los 8 resultados que se miran en el boxplot.

```python
Q3 = df2['capacidad'].quantile(0.75) # Tercer cuartil = Percentil 75
Q1 = df2['capacidad'].quantile(0.25) # Primer cuartil + Percentil 25

RIC = Q3 - Q1 # Obtener RIC

# Cálculo de los límites
superior = Q3 + (3 * RIC)
inferior = Q1 - (3* RIC)

outliers = df2[(df2['capacidad'] < inferior) | (df2['capacidad'] > superior)]
outliers
```
